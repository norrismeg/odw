```python
# -- download data
! wget https://dcc.ligo.org/public/0187/G2300818/001/challenge1.gwf

# -- for gwpy
from gwpy.timeseries import TimeSeries
gwpy_strain = TimeSeries.read('challenge1.gwf', channel="H1:CHALLENGE1")

# -- for pycbc
from pycbc import frame
pycbc_strain = frame.read_frame('challenge1.gwf', 'H1:CHALLENGE1')
```

```python
import numpy

import matplotlib.pyplot as plt
```

Sampling Rate and Duration of the Data:

```python
print('The sampling rate of the data is:',1/(pycbc_strain.delta_t),'Hz')
print('The duration of the data is:',len(pycbc_strain)/(1/(pycbc_strain.delta_t)),'seconds')
```

Plot of the data in the time-domain:

```python
ts_plot = pycbc_strain.plot()
plt.xlabel('Time (s)')
plt.ylabel('Strain')
```

Spectrogram and q-transform:

```python
specgram = gwpy_strain.spectrogram2(fftlength=4, overlap=2, window='hann') ** (1/2.)
plot_sp = specgram.plot()
ax = plot_sp.gca()
ax.set_yscale('log')
ax.set_ylim(10, 1400)
ax.colorbar(
    clim=(1e-24, 1e-20),
    norm="log",
    label=r"Strain noise [$1/\sqrt{\mathrm{Hz}}$]",
)
```

```python
strain = gwpy_strain.q_transform(frange=(30,500))
plot_q = strain.plot()
plot_q.colorbar(label="Normalised energy")

strain2 = gwpy_strain.q_transform(frange=(30, 500), outseg=(60,70),qrange=(4, 12))
plot_q2 = strain2.plot()
ax = plot_q2.gca()
ax.set_epoch(0)
ax.set_yscale('log')
ax.colorbar(label="Normalised energy")
```

The merger happens at 67 seconds.
